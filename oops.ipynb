```python
# oops--> object oriented programming
# oops is a way of writing programs using class and object
# oops is a programmming paradigm that organizes code using class and objject
# encapsulatio-->
# inhertance-->
# polymorphism-->
# abstraction-->
class student:
  def __init__(self,name,age):
    self.name=name
    self.age=age
  def info(self):
    print(self.name,self.age)
s1=student("ram",20)
s1.info()
```

```python
# A Decorator is a function that adds extra functionality to another function without modifying its original code.
def decorator_func(func):
    def wrapper():
        print("Before function call")
        func()
        print("After function call")
    return wrapper

@decorator_func
def greet():
    print("Hello")

greet()
```

```python
def func(a):
  def fuc():
    print("starting...")
    a()
    print("ending...")
  return fuc
@func
def hello():
  print("hello world")

hello()
```

```python
def decorator(func):

    def wrapper(*args, **kwargs):
        print("Starting...")
        func(*args, **kwargs)
        print("Finished...")

    return wrapper


class Employee:

    @decorator
    def show(self):
        print("Employee Details")

e = Employee()
e.show()
```

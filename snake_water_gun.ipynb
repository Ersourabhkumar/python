```python
import random
ls=["snake","water","gun"]
computer=random.choice(ls)
user=input("enter your choice")
if computer==user:
    print("tie")
elif computer=="snake" and user=="water":
    print("computer wins")
elif computer=="snake" and user=="gun":
    print("user wins")
```
